### Range in loop concept

```python
for i in range(6):  #start with 0 end with (6-1)
    print(i)
```

```python
for i in range(1,6): #(start with 1 end with (6-1)(stop)
    print(i)
```

```python
for i in range(4,9,2): #start with 4 and increment with step value that is 2 whether stop value (9-1)
    print(i)
```

```python
for i in range(6,1,-1):
    print(i)
```
